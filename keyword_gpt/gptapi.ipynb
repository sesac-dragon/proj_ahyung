````python
# 키워드가 들어간 문장 추출
import re

def keyword_sentences(text, keywords):
    # 문장 분리
    sentences = re.split(r'[.!?。\n]', text)

    result = []
    for s in sentences:
        for keyword in keywords:
            # 단어 단위 매칭: 주차장, 화장실 같은 단어를 정확히 찾도록 함
            if re.search(rf'\b{re.escape(keyword)}\b', s):
                result.append(s.strip())
                break
    return result

sample_text = df.iloc[1]['blogtext']
sentences = keyword_sentences(sample_text, keywords)

print("키워드 포함 문장:")
for s in sentences:
    print("-", s)
    
# 키워드 
import json
with open("./keyword_gpt/keywords.json", "r", encoding="utf-8") as f:
    keywords = json.load(f)
    if isinstance(keywords, dict):
        keywords = [kw for sublist in keywords.values() for kw in sublist]
````

## 최최종 코드 

````python
import pandas as pd
import pymysql
from db import load_env
import re
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain_openai import ChatOpenAI

# LLM 초기화
llm = ChatOpenAI(model_name='gpt-4o-mini', temperature=0.5)

# 데이터베이스에서 데이터 불러오기
def get_cafe_data():
    env = load_env(".env")
    conn = pymysql.connect(
        host=env["DB_HOST"],
        user=env["DB_USER"],
        password=env["DB_PASSWORD"],
        database=env["DB_DATABASE"],
        port=int(env["DB_PORT"]),
        charset="utf8mb4",
        cursorclass=pymysql.cursors.DictCursor,
    )
    query = "SELECT cafename, cafeaddress, latitude, longitude, blogdate, blogtext FROM tb_cafe ORDER BY blogdate DESC"
    with conn.cursor() as cur:
        cur.execute(query)
        result = cur.fetchall()
    conn.close()
    return pd.DataFrame(result)

# 키워드 포함 문장만 추출하는 함수
def keyword_sentences(text, keyword_list):
    sentences = re.split(r'[.!?。\n]', text)
    return [s.strip() for s in sentences if any(k in s for k in keyword_list)]

# GPT 프롬프트
template = """
너는 한국어 텍스트 분석 전문가야.
다음 문장들은({context}) 블로그 리뷰에서 추출된 내용이야. 

너가 각 문장에서 찾는 키워드는 뜨개카페와 관련된 키워드야
뜨개카페와 관련된 키워드가 있는지 없는지 앞 뒤 문장을 보고 문맥을 바탕으로 정확히 판단해서 정리해줘.
너가 생각했을 때 뽑아낸 키워드가 뜨개카페와 관련이 없으면 정리하지 마. 

예시를 참고해서 아래 형식으로 정리해줘.
키워드만 작성해서 정리해줘.

형식은 '키워드명: 실제 여부' 이렇게 정리해서 보여주면 돼.
(예시: 화장실, 없음 
주차장, 있음 등등 )
만약 모든 키워드의 실제 여부가 '없음'으로 나오면 키워드는 보여주지 말고 
'관련 정보가 없습니다.' 이것만 출력해줘
"""
prompt = PromptTemplate(input_variables=["context"], template=template)
chain = LLMChain(llm=llm, prompt=prompt)

# 키워드를 코드에 직접 정의
crochet_keywords = [
    "실종류","코바늘", "대바늘", "뜨개카페", "수세미", "원형뜨기", "모티브", "뜨개수업", "니팅", "손뜨개"
]
cafe_keywords = [
    "좌석", "조명", "화장실", "주차", "콘센트", "음악", "에어컨", "와이파이"
]

# 블로그 데이터 가져오기
df = get_cafe_data()
df = df[(df["latitude"].between(33.0, 38.7)) & (df["longitude"].between(124.5, 131.9))]

# 테스트용 샘플 한 개
sample_row = df.iloc[0]
sample_text = sample_row['blogtext']
print(f"카페이름: {sample_row['cafename']}")

# 키워드 문장 추출
crochet_sentences = keyword_sentences(sample_text, crochet_keywords)
cafe_sentences = keyword_sentences(sample_text, cafe_keywords)

# 분석 대상 문장 합치기
target_sentences = crochet_sentences + cafe_sentences
context = "\n".join(target_sentences)

# GPT 요청
if context.strip():
    print("GPT에게 보낼 문장들:")
    print(context)
    result = chain.invoke({"context": context})
    print("\n GPT 요약 결과:")
    print(result["text"])
else:
    print("키워드가 포함된 문장이 없습니다.")
````
