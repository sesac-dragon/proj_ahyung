````python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv()
openai_api_key = os.getenv('OPENAI_API_KEY')

llm = ChatOpenAI(model_name='gpt-4o-mini', temperature=0)
````

````python
# 뜨개 관련 키워드
# 카페 관련 키워드 
````

````python

template = """
너는 한국어를 기반으로 뜨개질 관련 키워드를 추출하는 텍스트 분석 전문가야.

뜨개질, 손뜨개, 코바늘, 대바늘, 털실, 뜨개카페, 실종류, 울 등과 같은 뜨개와 관련된 주요 키워드를 최대한 다양하게 추출해줘.

- 단일 단어 또는 짧은 구(phrase) 형태의 키워드로 구성해줘.
- 뜨개질 활동, 도구, 재료, 장소 등을 모두 포괄해서 작성해줘.
- 출력은 중복 없이, **한글 기준 30개 내외**로 **JSON 리스트 형식**으로 정리해줘.
"""

prompt = PromptTemplate(input_variables=[], template=template)
chain = LLMChain(llm=llm, prompt=prompt)
ch_res = chain.invoke({})
print(ch_res['text']) 
````

````python
template = """
너는 한국어를 기반으로 카페 관련 키워드를 추출하는 텍스트 분석 전문가야.

커피, 디저트, 인테리어, 분위기, 좌석, 베이커리 카페 등과 관련된 주요 키워드를 최대한 다양하게 추출해줘.

- 단일 단어 또는 짧은 구(phrase) 형태의 키워드로 구성해줘.
- 메뉴, 음료, 인테리어 요소, 화장실, 위치, 청결, 서비스, 방문 목적, 분위기, 손님 유형 등 다양한 범주를 포괄해서 작성해줘.
- 출력은 중복 없이, **한글 기준 30개 내외**로 **JSON 리스트 형식**으로 정리해줘.
"""

prompt = PromptTemplate(input_variables=[], template=template)
chain = LLMChain(llm=llm, prompt=prompt)

cafe_res = chain.invoke({})
print(cafe_res['text'])
````

````python
# 키워드 합치기
````

````python
import json
json_ch = json.loads(ch_res["text"].strip().replace("```json","").replace("```",""))
json_cafe = json.loads(cafe_res["text"].strip().replace("```json","").replace("```",""))
````

````python
keywords = json_ch + json_cafe
keywords
````

````python
# json 파일로 저장
all_keywords = {
    'crochet' : json_ch,
    'cafe' : json_cafe
}

with open("keywords.json", "w", encoding="utf-8") as f:
    json.dump(all_keywords, f, ensure_ascii=False)
````

````python
# 키워드가 들어간 문장 추출

import re

def keyword_sentences(text, keywords):
    # 문장 단위로 분리 (간단한 한글 기준)
    sentences = re.split(r'[.!?。\n]', text)
    # 키워드가 포함된 문장만 필터링
    return [s.strip() for s in sentences if any(keyword in s for keyword in keywords)]
````
